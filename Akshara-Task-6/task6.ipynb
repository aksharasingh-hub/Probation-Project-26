```python
df=df.apply(lambda x :x.str.strip() if x.dtype=="object" else x)
df
```

```python
df.columns
```

```python
df["Delivery_person_Age"] = df["Delivery_person_Age"].replace("NaN",pd.NA)
df["Delivery_person_Age"] = df["Delivery_person_Age"].fillna(df["Delivery_person_Age"].median())
df["Delivery_person_Age"] = df["Delivery_person_Age"].astype(int)
df
```

```python
df["Delivery_person_Ratings"] = df["Delivery_person_Ratings"].fillna(df["Delivery_person_Ratings"].mean())
```

```python
df["multiple_deliveries"]=df["multiple_deliveries"].fillna(df["multiple_deliveries"].mode()[0])
```

```python
df.duplicated().sum()
df=df.drop_duplicates()
```

```python
df.head(10)
```

```python
df["Time_taken(min)"] = (df["Time_taken(min)"].astype(str).str.replace("(min) ", "", regex=False).astype(int))
```

```python
df
```

```python
df.info()
```

```python
plt.boxplot(df["Time_taken(min)"], patch_artist=True, boxprops=dict(facecolor="red"))
plt.title("Outlier")
plt.ylabel("Time (Minutes)")
plt.show()
```

```python
a1 = df["Time_taken(min)"].quantile(0.25)
a3 = df["Time_taken(min)"].quantile(0.75)
iqr = a3 - a1
upper = a3 + 1.5 * iqr
df["Time_taken(min)"] = df["Time_taken(min)"].clip(0, upper)
```

```python
plt.boxplot(df["Time_taken(min)"],patch_artist=True, boxprops=dict(facecolor="lightgreen"))
plt.title("After Outlier Treatment")
plt.ylabel("Time (Minutes)")
plt.show()
```
