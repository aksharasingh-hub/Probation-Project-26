```python
num_cols = ["Delivery_person_Age","Delivery_person_Ratings","multiple_deliveries","Time_taken(min)"]
num_cols
```

```python
for col in num_cols:
    df[col]=pd.to_numeric(df[col],errors="coerce")
```

```python
df=df.replace('NaN',np.nan)
```

```python
df["Delivery_person_Age"] = df["Delivery_person_Age"].replace("NaN",pd.NA)
df["Delivery_person_Age"] = df["Delivery_person_Age"].fillna(df["Delivery_person_Age"].median())
df["Delivery_person_Age"] = df["Delivery_person_Age"].astype(int)
df
```

# Distribution of Delivery Time

```python
sns.histplot(df['Time_taken(min)'], bins=30, kde=True)
plt.title('Delivery Time')
plt.xlabel('Time Taken (min)')
plt.xticks(rotation=90)
```

The distribution is right-skewed, with most delivery times concentrated between 20–35 minutes. A small number of deliveries exceed 45 minutes, indicating the presence of outliers and occasional delayed deliveries

# Categorical(columns) Feature Analysis

```python

for col in text_cols:
        plt.figure(figsize=(6,4))
        sns.countplot(data=df, x=col)
        plt.title(col)
        plt.xticks(rotation=45)
plt.show()
```

# Delivery time vs Categorical columns

```python
fig, axes = plt.subplots(3, 2, figsize=(16, 14))
axes = axes.flatten()

for ax, col in zip(axes, text_cols):
    sns.boxplot(data=df, x=col, y="Time_taken(min)", ax=ax)
    ax.set_title(col)
    ax.set_ylabel("Time (min)")
    ax.set_ylim(10,50)
    ax.set_yticks([10,20,30,40,50])
    ax.tick_params(axis="x", rotation=25)
plt.tight_layout()
plt.show()
```

Weather: Cloudy and Fog weather have slightly higher delivery times than Sunny.
Road Traffic: Jam traffic shows the highest median delivery time, while Low traffic is the fastest.
Type of Order: Meal orders generally take a little longer than Snacks and Drinks.
Vehicle: Bicycle deliveries tend to take the longest; motorcycles are relatively faster.
Festival: Deliveries during festivals have a higher median time than non-festival days.
City: Metropolitan areas show longer delivery times compared to Urban c

# Delivery Time taken by Hour and Day

```python
df['Order_hour'] = df['Time_Orderd'].apply(lambda t: t.hour if pd.notnull(t) else np.nan)
plt.figure(figsize=(10, 5))
sns.boxplot(data=df, x='Order_hour', y='Time_taken(min)')
plt.title('Time Taken by Order Hour')
plt.show()
```

Evening peak hours take longer delivery time

```python

df["Order_Date"] = pd.to_datetime( df["Order_Date"],format="%d-%m-%Y",errors="coerce")
df["Order_day"] = df["Order_Date"].dt.day_name()
```

```python
df['Order_day'] = df['Order_Date'].dt.day_name()
plt.figure(figsize=(12, 6))
days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
sns.boxplot(data=df, x='Order_day', y='Time_taken(min)', order=days)
plt.title('Time Taken by Day of the Week')
plt.xticks(rotation=30)
plt.show()
```

Evening and weekend orders generally take longer to deliver. Higher demand during these periods may increase delivery time.

# Correlation between numeric columns

```python
numeric= df[['Delivery_person_Age', 'Delivery_person_Ratings','Vehicle_condition','multiple_deliveries', 'Time_taken(min)']]

plt.figure(figsize=(8, 6))
sns.heatmap(numeric.corr(), annot=True, fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()
```

Multiple deliveries have the strongest positive correlation with delivery time (0.39), so more deliveries generally lead to longer delivery times.
Delivery person ratings show a negative correlation (−0.24), indicating higher-rated delivery partners tend to deliver faster.
Age has only a weak positive relationship (0.29), while vehicle condition has a very weak effect (−0.04).
